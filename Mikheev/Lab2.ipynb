```python
import random

# Функция для создания омофонной таблицы
def create_homophonic_table():
    # Определяем символы и их омофоны
    homophones = {
        'A': ['1', '2', '3'],
        'B': ['4', '5'],
        'C': ['6', '7'],
        'D': ['8', '9'],
        'E': ['0', 'A', 'B'],
        'F': ['C', 'D'],
        'G': ['E', 'F'],
        'H': ['G', 'H'],
        'I': ['I', 'J'],
        'K': ['K', 'L'],
        'L': ['M', 'N'],
        'M': ['O', 'P'],
        'N': ['Q', 'R'],
        'O': ['S', 'T'],
        'P': ['U', 'V'],
        'Q': ['W', 'X'],
        'R': ['Y', 'Z'],
        'S': ['!', '@'],
        'T': ['#', '$'],
        'U': ['%', '^'],
        'V': ['&', '*'],
        'W': ['(', ')'],
        'X': ['-', '='],
        'Y': ['+', '{'],
        'Z': ['}', '['],
        ' ': [' ', '_']  # Пробелы могут оставаться
    }
    return homophones

# Функция для шифрования текста
def encrypt(text, homophones):
    encrypted_text = ''
    for char in text.upper():
        if char in homophones:
            encrypted_text += random.choice(homophones[char])  # Выбираем случайный омофон
        else:
            encrypted_text += char  # Если символ не в таблице, оставляем его без изменений
    return encrypted_text

# Функция для дешифрования текста
def decrypt(encrypted_text, homophones):
    reverse_homophones = {v: k for k, vals in homophones.items() for v in vals}
    decrypted_text = ''
    for char in encrypted_text:
        if char in reverse_homophones:
            decrypted_text += reverse_homophones[char]  # Заменяем омофон на оригинальный символ
        else:
            decrypted_text += char  # Если символ не в таблице, оставляем его без изменений
    return decrypted_text

# Создание файла input.txt с примерным содержимым
with open('input.txt', 'w', encoding='utf-8') as f:
    f.write("Это пример текста для шифрования с использованием метода омофонов.")

# Создание омофонной таблицы
homophones = create_homophonic_table()

# Шифрование файла
input_file = 'input.txt'
output_file_encrypted = 'encrypted.txt'
output_file_decrypted = 'decrypted.txt'

# Чтение содержимого файла и шифрование
with open(input_file, 'r', encoding='utf-8') as f:
    content = f.read()

encrypted_content = encrypt(content, homophones)

# Запись зашифрованного текста в файл
with open(output_file_encrypted, 'w', encoding='utf-8') as f:
    f.write(encrypted_content)

# Дешифрование
decrypted_content = decrypt(encrypted_content, homophones)

# Запись расшифрованного текста в файл
with open(output_file_decrypted, 'w', encoding='utf-8') as f:
    f.write(decrypted_content)

print('Шифрование и дешифрование завершены.')
```
