```python
def rearrange_string(input_string):
    # Разбиваем строку на группы по 4 символа
    groups = [input_string[i:i+4] for i in range(0, len(input_string), 4)]

    # Функция для перестановки символов в группе
    def rearrange_group(group):
        # Пример перестановки: просто перевернем группу
        return group[::-1]  # Можно изменить на любую другую логику перестановки

    # Применяем перестановку к каждой группе
    rearranged_groups = [rearrange_group(group) for group in groups]

    # Объединяем группы обратно в строку
    return ''.join(rearranged_groups)

# Пример использования
input_string = "abcdefghijklmno"
result = rearrange_string(input_string)
print(result)  # Вывод: dcba hgfe lkji nm
```

```python
def encrypt_route_transposition(input_string):
    # Определяем размеры таблицы
    rows = 5
    cols = 8

    # Заполняем таблицу символами из входной строки
    table = [['' for _ in range(cols)] for _ in range(rows)]

    index = 0
    for r in range(rows):
        for c in range(cols):
            if index < len(input_string):
                table[r][c] = input_string[index]
                index += 1

    # Определяем маршрут для считывания
    # Например, считываем по столбцам
    encrypted_string = ''
    for c in range(cols):
        for r in range(rows):
            if table[r][c]:  # Проверяем, что ячейка не пустая
                encrypted_string += table[r][c]

    return encrypted_string

# Пример использования
input_string = "Это пример текста для шифрования"
# Убедимся, что строка достаточно длинная для заполнения таблицы
input_string = input_string.ljust(40)  # Дополняем пробелами до 40 символов
result = encrypt_route_transposition(input_string)
print(result)
```

```python
def encrypt_scytale(input_string, num_columns):
    # Удаляем пробелы и приводим строку к верхнему регистру
    input_string = input_string.replace(" ", "").upper()

    # Вычисляем количество строк, необходимых для заполнения таблицы
    num_rows = (len(input_string) + num_columns - 1) // num_columns

    # Создаем таблицу
    table = [['' for _ in range(num_columns)] for _ in range(num_rows)]

    # Заполняем таблицу символами
    index = 0
    for r in range(num_rows):
        for c in range(num_columns):
            if index < len(input_string):
                table[r][c] = input_string[index]
                index += 1

    # Считываем по столбцам для шифрования
    encrypted_string = ''
    for c in range(num_columns):
        for r in range(num_rows):
            if table[r][c]:  # Проверяем, что ячейка не пустая
                encrypted_string += table[r][c]

    return encrypted_string

def decrypt_scytale(encrypted_string, num_columns):
    # Вычисляем количество строк
    num_rows = (len(encrypted_string) + num_columns - 1) // num_columns

    # Создаем таблицу
    table = [['' for _ in range(num_columns)] for _ in range(num_rows)]

    # Заполняем таблицу по столбцам
    index = 0
    for c in range(num_columns):
        for r in range(num_rows):
            if index < len(encrypted_string):
                table[r][c] = encrypted_string[index]
                index += 1

    # Считываем по строкам для дешифрования
    decrypted_string = ''
    for r in range(num_rows):
        for c in range(num_columns):
            if table[r][c]:  # Проверяем, что ячейка не пустая
                decrypted_string += table[r][c]

    return decrypted_string

# Пример использования
input_string = "Это пример текста для шифрования"
num_columns = 8

# Шифрование
encrypted = encrypt_scytale(input_string, num_columns)
print("Зашифрованный текст:", encrypted)

# Дешифрование
decrypted = decrypt_scytale(encrypted, num_columns)
print("Расшифрованный текст:", decrypted)
```

```python
def rotate_grid(grid):
    """Поворачивает решетку на 90 градусов по часовой стрелке."""
    return [list(reversed(col)) for col in zip(*grid)]

def encrypt(text):
    """Шифрует текст с использованием поворотной решетки 4x4."""
    # Убираем пробелы и ограничиваем длину текста
    text = text.replace(" ", "")[:16]

    # Заполняем решетку
    grid = [[''] * 4 for _ in range(4)]
    index = 0
    for i in range(4):
        for j in range(4):
            if index < len(text):
                grid[i][j] = text[index]
                index += 1

    # Поворачиваем решетку
    rotated_grid = rotate_grid(grid)

    # Читаем зашифрованный текст
    encrypted_text = ''.join(''.join(row) for row in rotated_grid)
    return encrypted_text

# Пример использования
original_text = "HELLO WORLD"
encrypted = encrypt(original_text)
print("Зашифрованный текст:", encrypted)
```

```python
def create_table(text, rows, cols):
    """Создает таблицу и заполняет ее текстом."""
    table = [[''] * cols for _ in range(rows)]
    index = 0
    for i in range(rows):
        for j in range(cols):
            if index < len(text):
                table[i][j] = text[index]
                index += 1
    return table

def permute_rows(table, row_order):
    """Переставляет строки таблицы по заданному порядку."""
    return [table[i] for i in row_order]

def permute_columns(table, col_order):
    """Переставляет столбцы таблицы по заданному порядку."""
    return [[row[i] for i in col_order] for row in table]

def encrypt(text, row_order, col_order):
    """Шифрует текст с использованием двойной перестановки."""
    # Убираем пробелы и ограничиваем длину текста
    text = text.replace(" ", "")[:16]

    # Создаем таблицу 4x4
    table = create_table(text, 4, 4)

    # Первая перестановка (строки)
    permuted_rows = permute_rows(table, row_order)

    # Вторая перестановка (столбцы)
    permuted_table = permute_columns(permuted_rows, col_order)

    # Читаем зашифрованный текст
    encrypted_text = ''.join(''.join(row) for row in permuted_table)
    return encrypted_text

# Пример использования
original_text = "HELLO WORLD"
row_order = [2, 0, 3, 1]  # Порядок перестановки строк
col_order = [3, 1, 0, 2]  # Порядок перестановки столбцов

encrypted = encrypt(original_text, row_order, col_order)
print("Зашифрованный текст:", encrypted)
```
