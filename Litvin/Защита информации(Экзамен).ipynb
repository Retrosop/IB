Первая лабораторная, дешифрование, метод перестановки битов

```python
from cryptography.fernet import Fernet

# Генерация ключа
key = Fernet.generate_key()
cipher = Fernet(key)

# Исходное сообщение
original_message = "Hello, World!"

# Шифрование
encrypted_message = cipher.encrypt(original_message.encode())
print(f"Зашифрованное сообщение: {encrypted_message}")

# Дешифрование
decrypted_message = cipher.decrypt(encrypted_message).decode()
print(f"Расшифрованное сообщение: {decrypted_message}")
```

Вторая работа, шифрование, система омофонов

```python
import random

# Создание таблицы замен
def create_substitution_table():
    substitution_table = {
        'A': ['1', '2', '3'],
        'B': ['4', '5'],
        'C': ['6', '7', '8'],
        # Добавьте остальные буквы
    }
    return substitution_table

# Шифрование
def encrypt(message, substitution_table):
    encrypted_message = []
    for char in message:
        if char in substitution_table:
            encrypted_char = random.choice(substitution_table[char])
            encrypted_message.append(encrypted_char)
        else:
            encrypted_message.append(char)  # Неизвестные символы остаются без изменений
    return ''.join(encrypted_message)

# Дешифрование (требует знания оригинальной таблицы замен)
def decrypt(encrypted_message, substitution_table):
    reverse_table = {v: k for k, values in substitution_table.items() for v in values}
    decrypted_message = []
    for char in encrypted_message:
        if char in reverse_table:
            decrypted_char = reverse_table[char]
            decrypted_message.append(decrypted_char)
        else:
            decrypted_message.append(char)  # Неизвестные символы остаются без изменений
    return ''.join(decrypted_message)

# Пример использования
substitution_table = create_substitution_table()
original_message = "ABAC"
encrypted = encrypt(original_message, substitution_table)
decrypted = decrypt(encrypted, substitution_table)

print(f"Оригинальное сообщение: {original_message}")
print(f"Зашифрованное сообщение: {encrypted}")
print(f"Расшифрованное сообщение: {decrypted}")
```

Третья работа, Зашифровать и расшифровать текст, находящийся в файле с именем Source.txt. Закодированный текст сохранить в файле с именем Coded.txt, расшифрованный текст записать в файл DeCoded.txt.
линейный конгруэнтный датчик.

```python
import os
import random

# Линейный конгруэнтный генератор
def linear_congruential_generator(seed, a=1664525, c=1013904223, m=2**32):
    while True:
        seed = (a * seed + c) % m
        yield seed

# Шифрование текста
def encrypt(text, seed):
    lcg = linear_congruential_generator(seed)
    encrypted_text = ''.join(chr((ord(char) + next(lcg)) % 256) for char in text)
    return encrypted_text

# Дешифрование текста
def decrypt(encrypted_text, seed):
    lcg = linear_congruential_generator(seed)
    decrypted_text = ''.join(chr((ord(char) - next(lcg)) % 256) for char in encrypted_text)
    return decrypted_text

# Основная функция
def main():
    # Путь к файлам
    source_file_path = r'C:\Users\Kilen\Desktop\Source.txt'
    coded_file_path = r'C:\Users\Kilen\Desktop\Coded.txt'
    decoded_file_path = r'C:\Users\Kilen\Desktop\DeCoded.txt'

    # Проверка существования файла Source.txt
    if not os.path.exists(source_file_path):
        # Создание файла с примерным текстом, если он не существует
        with open(source_file_path, 'w', encoding='utf-8') as file:
            file.write("Это пример текста для шифрования.")  # Пример текста

    # Чтение исходного текста
    with open(source_file_path, 'r', encoding='utf-8') as file:
        original_text = file.read()

    # Установка начального значения для ЛКГ
    seed = 12345  # Можно использовать любое значение

    # Шифрование
    coded_text = encrypt(original_text, seed)
    with open(coded_file_path, 'w', encoding='utf-8') as file:
        file.write(coded_text)

    # Дешифрование
    decoded_text = decrypt(coded_text, seed)
    with open(decoded_file_path, 'w', encoding='utf-8') as file:
        file.write(decoded_text)

    print("Шифрование и дешифрование завершены.")

    # Вывод содержимого файлов на экран
    print("\nСодержимое Coded.txt:")
    with open(coded_file_path, 'r', encoding='utf-8') as file:
        print(file.read())

    print("\nСодержимое DeCoded.txt:")
    with open(decoded_file_path, 'r', encoding='utf-8') as file:
        print(file.read())

# Запуск программы
if __name__ == "__main__":
    main()
```

Четвертая работа, Смоделировать декодер с бесконечной последовательностью ключей формируемых  с помощью аддитивного псевдо-случайного датчика случайных чисел.

```python
class AdditivePRNG:
    def __init__(self, x0, x1, m):
        self.x0 = x0
        self.x1 = x1
        self.m = m

    def next(self):
        x_next = (self.x0 + self.x1) % self.m
        self.x0, self.x1 = self.x1, x_next
        return x_next

# Пример использования
prng = AdditivePRNG(1, 2, 100)  # Начальные значения и модуль
keys = [prng.next() for _ in range(10)]  # Генерация 10 ключей
print(keys)
```

Пятая работая, Напишите формулу для расшифровки полученного сообщения.

```python
def decode(data, keys):
    return ''.join(chr(ord(c) ^ k) for c, k in zip(data, keys))

# Пример использования
encrypted_message = "зашифрованные данные"  # Замените на ваше зашифрованное сообщение
keys = [1, 2, 3, 4, 5]  # Замените на ваши сгенерированные ключи
decoded_message = decode(encrypted_message, keys)
print(decoded_message)
```

Шестая работа, Добавление пароля в начало зашифрованного файла.

```python
from cryptography.fernet import Fernet

# Генерация ключа и создание объекта шифрования
key = Fernet.generate_key()
cipher = Fernet(key)

# Исходные данные
data = "Это секретное сообщение"
encrypted_data = cipher.encrypt(data.encode())

# Пароль для доступа
password = "my_secure_password"

# Создание зашифрованного файла с паролем в начале
with open("encrypted_file.bin", "wb") as file:
    # Запись пароля
    file.write(password.encode() + b'\n')
    # Запись зашифрованных данных
    file.write(encrypted_data)

print("Файл успешно зашифрован и сохранен.")

# Чтение зашифрованного файла
with open("encrypted_file.bin", "rb") as file:
    # Чтение пароля
    password = file.readline().strip().decode()
    # Чтение зашифрованных данных
    encrypted_data = file.read()

# Расшифровка данных
decrypted_data = cipher.decrypt(encrypted_data).decode()

print("Пароль:", password)
print("Расшифрованное сообщение:", decrypted_data)
```
